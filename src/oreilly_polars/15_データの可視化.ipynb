# データの可視化

```python
import os

import hvplot
import polars as pl
```

## 定数定義

```python
DATA_PAR_PATH = os.path.join('..','..','data')
INPUT_PARQUETS_PATH_TAXI = os.path.join(DATA_PAR_PATH,'TAXI','*.parquet')
```

## データの読み込み

```python
taxis = pl.read_parquet(INPUT_PARQUETS_PATH_TAXI)
taxis
```

```python
taxis.filter(
    (pl.col('VendorID') == 5)
)
```

```python
taxis_ = (
    taxis
    .filter(pl.col('VendorID') == 5)
    .select(
        pl.col('trip_distance'),
        abs(pl.col('tpep_dropoff_datetime') - pl.col('tpep_pickup_datetime')).cast(pl.Float32).alias('duration'),
        pl.col('tip_amount')
    )
)

taxis_
```

```python
taxis_.plot.scatter(x='trip_distance', y='duration', color='tip_amount')
```
