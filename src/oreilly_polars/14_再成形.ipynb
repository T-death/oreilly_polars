## 横に長い / 縦に長いDataFrame

```python
df = pl.DataFrame({
    'studnet': ['Alice', 'Bob', 'Charlie'],
    'math': [85, 78, 92],
    'science': [90, 82, 85],
    'history': [88, 80, 87]
})

df
```

```python
df = pl.DataFrame({
    'student': ['Alice','Alice','Alice', 'Bob','Bob','Bob', 'Charlie', 'Charlie', 'Charlie'],
    'subject': ['Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History'],
    'grade': [85, 90, 88, 78, 82, 80, 92, 85, 87]
})

df
```

```python
df.pivot(index='student', on='subject', values='grade')
```

```python
df = pl.DataFrame({
    'student': ['Alice','Alice','Alice', 'Alice','Alice','Alice', 'Bob','Bob','Bob', 'Bob','Bob','Bob'],
    'subject': ['Math', 'Math', 'Math', 'Science', 'Science', 'Science', 'Math', 'Math', 'Math', 'Science', 'Science', 'Science'],
    'grade': [85, 88, 85, 60, 66, 63, 51, 79, 62, 82, 85, 82]
})

df
```

```python
df.pivot(
    index='student',
    on='subject',
    values='grade',
    aggregate_function='mean'
)
```

`aggregate_function`を指定することによって、意味が重複しているデータの研鑽処理が実行できるのか

```python
df.pivot(
    index='student',
    on='subject',
    values='grade',
    aggregate_function=pl.element().max() - pl.element().min()
)
```

```python
df = pl.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie'],
    'math': [85, 78, 92],
    'science': [90, 82, 85],
    'history': [88, 80, 87]
})

df
```

```python
df.unpivot(
    index=['student'],
    on=['math', 'science', 'history'],
    variable_name='subject',
    value_name='grade'
)
```

前のpolarsだと`.melt()`だったけれど、今のバージョンでは`.unpivot()`になったみたい。  
meltでもなんとなくわかるけれど、unpivotの方がよりわかりやすいかも

```python
df = pl.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'class': ['Math101', 'Math101', 'Math101', 'Math102', 'Math102', 'Math102'],
    'age': [20, 21, 22, 20, 21, 22],
    'semester': ['Fall', 'Fall', 'Fall', 'Spring', 'Spring', 'Spring'],
    'math': [85, 78, 92, 88, 79, 95],
    'science': [90, 82, 85, 92, 81, 87],
    'history': [88, 80, 87, 85, 82, 89]
})

df
```

```python
df.unpivot(
    index=['student', 'class', 'age', 'semester'],
    on=['math', 'science', 'history'],
    variable_name='subject',
    value_name='grade'
)
```

## 分割する

```python
df = pl.DataFrame({
    'OrderID': [1, 2, 3, 4, 5, 6],
    'Product': ['A', 'B', 'A', 'C', 'B', 'A'],
    'Quantity': [10, 5, 8, 7, 3, 12],
    'Region': ['North', 'South', 'North', 'West', 'South', 'West']
})
```

```python
df.partition_by('Region')
```

`.partition_by()`で、指定したカラムのグループごとにテーブルを分割してくれるのか。  
`.rolling()`だと元のテーブル構造を崩さずに、グループ別の情報を追加していたので、どっちを使うかはケースバイケース

```python
df.partition_by('Region', include_key=False)
```

```python
dfs = df.partition_by(['Region'], as_dict=True)
dfs
```

これ面白いメソッドだなぁ。  
結構使い所ありそう！

```python
dfs[('North',)]
```
